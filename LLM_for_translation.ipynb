#Translate text

```python
import os, json
from openai import OpenAI
from google.colab import userdata


# 1. Standard OpenAI Client
openai_client = OpenAI(
    api_key=userdata.get('chatgpt_api_key2')
)

# 2. DeepSeek Client (Using the same OpenAI SDK)
deepseek_client = OpenAI(
    api_key= userdata.get('deepseek_api_key'),
    base_url="https://api.deepseek.com" # Point to DeepSeek's server
)


text_to_translate="The U.S. and China agreed to lower tariffs on $30 billion worth of goods each country imports from the other and launch new talks on artificial intelligence (AI) following President Trump’s summit with Chinese President Xi Jinping."

current_model="gpt-4.1-nano"
current_messages=[{"role": "system", "content": "translate the input text to Spanish. output the translation only"},
    {"role": "user", "content": text_to_translate}]


gpt_response = openai_client.chat.completions.create(
    model=current_model,
    messages=current_messages
)


gpt_out=json.loads(gpt_response.model_dump_json())
gpt_out

translated_text=gpt_out["choices"][0]["message"]["content"]
cost=calculate_api_call_cost(gpt_out,current_model,pricing_dict_selected_models)
print("translated_text:",translated_text)
print("cost:",cost)
```

#Translate using multiple models

```python
openai_models=["gpt-5.4-mini","gpt-5.4-nano","gpt-5.2","gpt-5.1","gpt-4.1","gpt-4.1-mini","gpt-4.1-nano","gpt-4o"]
deepseek_models=["deepseek-flash","deepseek-v4-pro"]

text_to_translate="The U.S. and China agreed to lower tariffs on $30 billion worth of goods each country imports from the other and launch new talks on artificial intelligence (AI) following President Trump’s summit with Chinese President Xi Jinping."

current_messages=[{"role": "system", "content": "translate the input text to Spanish. output the translation only"},
    {"role": "user", "content": text_to_translate}]


for current_model in openai_models:
  gpt_response = openai_client.chat.completions.create(
      model=current_model,
      messages=current_messages
  )
  gpt_out=json.loads(gpt_response.model_dump_json())
  translated_text=gpt_out["choices"][0]["message"]["content"]
  cost=calculate_api_call_cost(gpt_out,current_model,pricing_dict_selected_models)
  print(current_model)
  print("translated_text:",translated_text)
  print("cost:",cost)


for current_model in deepseek_models:
  ds_response = deepseek_client.chat.completions.create(
      model=current_model,
      messages=current_messages,
      extra_body={"thinking": {"type": "disabled"},"reasoning_effort":"none"}
  )
  gpt_out=json.loads(gpt_response.model_dump_json())
  translated_text=gpt_out["choices"][0]["message"]["content"]
  cost=calculate_api_call_cost(gpt_out,current_model,pricing_dict_selected_models)
  print(current_model)
  print("translated_text:",translated_text)
  print("cost:",cost)

```
